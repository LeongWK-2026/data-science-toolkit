```python
import numpy as np
```

```python
#Question 1: Create a numpy array of size 2 x 2
array_2x2 = np.array ([
    [1,2],
    [3,4]
])
print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#Question 2: Create a random numpy array of size 3 x 4
random3x4 = np.random.rand(3,4)
print("random 3x4 array")
print(random3x4)
```

```python
#Question 3: Create a numpy array of ones and a numpy array of zeroes
array_ones = np.ones((3,3))
array_zeros = np.zeros((3,3))

print("Array of ones")
print(array_ones)

print("Array of zeros")
print(array_zeros)
```

```python
#Question 4: Create a random numpy array of size 6x3 and transpose it
arr_6x3 = np.random.rand(6,3)
transposed_arr = np.transpose(arr_6x3)

print("Original 6x3 Array Shape:", arr_6x3.shape)
print("Original 6x3 Array: ", arr_6x3)
print("\nTransposed Array Shape:", transposed_arr.shape)
print("Transposed Array:", transposed_arr)
```

```python
#Question 5: Create two random numpy array of size n x p and p x m and then compute the dot product

array_A = np.random.rand(3,4)
array_B = np.random.rand(4,2)

dot_product = np.dot(array_A, array_B)

print(f"Array A shape: {array_A.shape}")
print("Array A: ", array_A)
print(f"\nArray B shape: {array_B.shape}")
print("Array B : ", array_B)
print(f"\nDot product shape: {dot_product.shape}")
print("Dot Product:")
print(dot_product)
```
